# HDMI 实时 Sobel（HDMI 版 overlay）

本 notebook 做的事和 `hdmi_sobel.py --seconds 10` 相同：加载 overlay、配置 HDMI 输入输出、
启动管线。

**注意**：这一版视频流绕过 VDMA 直通内核，所以 `hdmi_in.readframe()` 不可用 ——
要看处理结果，需要把显示器或 HDMI 采集卡接到板子的 HDMI OUT。

运行前确认本目录下有 `sobel_hdmi.bit`、`sobel_hdmi.hwh`。

```python
import os
import pynq
print('PYNQ', pynq.__version__)
for f in ['sobel_hdmi.bit', 'sobel_hdmi.hwh']:
    print(f, 'present:', os.path.exists(f))
```

```python
import subprocess, sys
rc = subprocess.call([sys.executable, 'hdmi_sobel.py', '--seconds', '15'])
print('return code:', rc)
```

## 结果怎么看

- `detected input mode: 1280x720 @ 60.00 Hz ...` → HDMI 输入已锁定，源接好了；
- `pipeline running - HDMI OUT should now show the Sobel edge image` → 输出管线已启动；
- 屏幕/采集卡上看到的应该是**黑白边缘图**；如果还是原图，检查 `.bit` 是不是这次构建的。

如果报 `no HDMI input signal detected`：确认笔记本处于"扩展/复制"模式，且分辨率是
720p60（1080p 也能跑，但本项目按 720p 验证）。
