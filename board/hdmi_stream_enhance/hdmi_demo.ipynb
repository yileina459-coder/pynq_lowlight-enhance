# 演示：逐个切换预设（录视频用这一份）

运行下面两格之后，**采集卡窗口里的画面会每 4 秒变一次**。
改 `HOLD` 可以调整每个预设停留的秒数，改 `ORDER` 可以调整顺序。

先运行第一格加载，再运行第二格开始循环；想停下就点工具栏的 ⏹。


```python
import time
from pynq import Overlay
import hdmi_live as hl

ol = Overlay("enhance_hdmi.bit")
video = ol.video
cfg = hl.find(ol, "gpio_enh_cfg")
lat = hl.find(ol, "gpio_enh_lat")
print("overlay loaded, IPS:", len(ol.ip_dict))
print("mode:", video.hdmi_in.mode)
```

```python
HOLD = 4                     # 每个预设停留的秒数
ORDER = ["pass", "clarity", "document", "night", "invert", "binarize", "split"]
ROUNDS = 1                   # 想循环多轮就调大

for r in range(ROUNDS):
    for name in ORDER:
        hl.write_preset(cfg, name)
        n = hl.read_latency(lat)
        stamp = time.strftime("%H:%M:%S")
        lat_txt = "(no stamps yet)" if n is None else f"{n} pixel clocks (~{n / 142.857 / 1000:.3f} us)"
        print(f"[{stamp}] {name:<9} latency {lat_txt}")
        time.sleep(HOLD)
print("done")
```
