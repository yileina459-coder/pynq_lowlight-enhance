# 自检：overlay 能不能加载 + HDMI 输入有没有锁定

**运行方式**：菜单 `Run` → `Run All Cells`，或者按 `Shift+Enter` 逐格运行。

看两件事：
1. `IPS:` 应该是一个大于 0 的数（说明位流加载成功）；
2. `mode:` 应该是 `1280x720@60` 之类的值。

**如果 `mode: None`**：线 A 没插好、笔记本没往外送信号，或者分辨率不是 PYNQ 能锁的模式。
先把笔记本外接屏设成 1280×720，再重新运行一次这一格。


```python
import time
from pynq import Overlay
import hdmi_live as hl

ol = Overlay("enhance_hdmi.bit")
video = ol.video
cfg = hl.find(ol, "gpio_enh_cfg")
lat = hl.find(ol, "gpio_enh_lat")
print("overlay loaded, IPS:", len(ol.ip_dict))
print("mode:", video.hdmi_in.mode)
```

## 写一个预设并读回延迟

`latency` 是内核流水线延迟（两个时间戳相减）。
如果它是 `None`，说明上电后还没有像素流过——也就是输入还没锁定，先解决上一格。


```python
hl.write_preset(cfg, "document")
print("preset document applied")
n = hl.read_latency(lat)
print("latency:", n, "pixel clocks" if n is not None else "(not measured yet)")
if n is not None:
    print(f"         ~{n / 142.857 / 1000:.3f} us at 142.857 MHz")
```
