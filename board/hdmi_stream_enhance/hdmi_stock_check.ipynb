# 对照实验：原厂 base overlay 能不能看到 HDMI 输入

用来分清两种完全不同的故障：

* 这一格也报 `None` → **板子这一侧（口/线/EDID）有问题**，与我们的 overlay 无关；
* 这一格报 `1280x720@60` → 原厂能收到，**是我们的 overlay 弄坏了什么**，需要我修。

运行前把笔记本 HDMI 插在板子的 **HDMI IN** 上。


```python
from pynq import Overlay

STOCK = "/usr/local/share/pynq-venv/lib/python3.10/site-packages/pynq/overlays/base/base.bit"
ol = Overlay(STOCK)
print("stock overlay loaded, IPs:", len(ol.ip_dict))
print("mode:", ol.video.hdmi_in.mode)
```

## 如果第一次是 None，插拔一次线再跑这一格

EDID 握手发生在插线的那一刻；如果线在位流加载前就插好了，有些笔记本不会重试。


```python
import time

for i in range(10):
    m = ol.video.hdmi_in.mode
    print(i, "mode:", m)
    if m is not None:
        print("SEES SOURCE ->", m.width, "x", m.height, "@", m.frame_rate)
        break
    time.sleep(1.0)
```
